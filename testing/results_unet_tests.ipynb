This jupyter notebook gather all metrics from the U-Net tests in a single place
***

```python
import json
from pathlib import Path
import pandas as pd
```

```python
# Get the directory where THIS script is located
base_path = Path.cwd()
base_path = base_path.parent  # Move one level up to the project root
```

```python
models_path = base_path / "unet_tests"
output_path = base_path / "unet_tests" / "models_summary.csv"
```

```python
# Load JSON files automatically

metrics_data = []

for metrics_file in models_path.glob("metrics_data_*.json"):
    if metrics_file.exists():
        with open(metrics_file, "r") as f:
            metrics_data.append(json.load(f))
```

```python
# Create DataFrames directly from the JSON dictionaries

metrics_df = pd.DataFrame(metrics_data)
```

```python
# Remove confusion matrix

metrics_df = metrics_df.drop(columns=["confusion_matrix"], errors="ignore")
```

```python
# Use model names as columns

metrics_df = metrics_df.set_index("Model").T
```

```python
# Check result
print(f"\nMerged DataFrame shape: {metrics_df.shape}")
print(metrics_df)
```

```python
# Save

metrics_df.to_csv(output_path, index=True)
print(f"\nSaved to: {output_path}")
```
